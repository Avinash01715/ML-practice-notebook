```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
# creating dataset with 5 features or dimensions
X, y = make_blobs(n_samples=500, n_features=5, centers=3, cluster_std=1.5,random_state=40)
```

```python
X
```

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
#pca model to convert 5 dimensions to 2 dimensions/features
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)
```

```python
X_pca
```

```python
df_pca = pd.DataFrame(X_pca, columns=['PC1', 'PC2'])
df_pca['label'] = y
```

```python
sns.scatterplot(data=df_pca, x='PC1', y='PC2', hue='label', palette='Set2')
```
