```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
df = sns.load_dataset('iris')
df.head()
```

```python
X = df.drop('species', axis = 1)
y = df['species']
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

KNN

```python
from sklearn.neighbors import KNeighborsClassifier

model_knn = KNeighborsClassifier(n_neighbors = 5)
```

```python
model_knn.fit(X_train, y_train)
```

```python
model_knn.score(X_test,y_test)
```

SVM

```python
from sklearn.svm import SVC

model_svc = SVC(C = 30, kernel = 'rbf', gamma= 'auto')
```

```python
model_svc.fit(X_train, y_train)
```

```python
model_svc.score(X_test,y_test)
```

**Grid search cv** ---> here no need of train test split

for SVM

```python
from sklearn.model_selection import GridSearchCV
```

```python
classifier_svc = GridSearchCV((model_svc),{
    'C': [1,10,20,30],
    'kernel': ['rbf', 'linear'],
}, cv = 5 , return_train_score= False)
```

```python
classifier_svc.fit(X,y)
```

```python
results = pd.DataFrame(classifier_svc.cv_results_)
results
```

```python
results[['param_C','param_kernel','mean_test_score']]
```

for Knn

```python
classifier_knn = GridSearchCV((model_knn),{
    'n_neighbors': [3,5,7,9,11,13],
    'weights': ['uniform','distance'],
    'metric': ['euclidean', 'manhattan', 'chebyshev', 'hamming'],
},cv = 5, return_train_score = False)
```

```python
classifier_knn.fit(X,y)
```

```python
result_knn = pd.DataFrame(classifier_knn.cv_results_)
result_knn.columns
```

```python
result_knn[['param_metric', 'param_n_neighbors', 'param_weights','mean_test_score']]
```
